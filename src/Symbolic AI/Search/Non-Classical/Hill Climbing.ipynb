# Hill Climbing Algorithm

Hill Climbing is a **local search algorithm** used for optimization problems.

It starts from an initial state and repeatedly moves to a neighboring state
with a better evaluation value.

### Main Idea
At each step:
- evaluate the current state
- generate neighboring states
- move to the best neighbor if it improves the current value

The algorithm stops when:
- no neighbor is better than the current state

This point is called a **local optimum**.

### Characteristics
- Simple and fast
- Uses very little memory
- Does not guarantee the global optimum
- Can get stuck in:
  - local maxima
  - plateaus
  - ridges


### Step 1: Define the Objective Function

We define a simple mathematical function to maximize.

In this example:

$$
f(x) = -x^2 + 6x
$$

This is a parabola opening downward, so it has a maximum point.
Hill Climbing will try to move toward higher values of this function.


```python
def objective_function(x):
    return -x**2 + 6*x
```

### Step 2: Define Neighbor Generation

For a one-dimensional problem, the neighbors of a point `x`
can be points slightly to the left or right.

We use a step size of `1`.


```python
def get_neighbors(x, step_size=1):
    return [x - step_size, x + step_size]
```

### Step 3: Implement Hill Climbing

This function:
- starts from an initial point
- checks its neighbors
- moves to the neighbor with the highest value if it is better
- stops when no improvement is possible


```python
def hill_climbing(start, step_size=1, max_iterations=100):
    current = start
    current_value = objective_function(current)

    for iteration in range(max_iterations):
        neighbors = get_neighbors(current, step_size)
        neighbor_values = [objective_function(n) for n in neighbors]

        best_neighbor = neighbors[neighbor_values.index(max(neighbor_values))]
        best_neighbor_value = objective_function(best_neighbor)

        print(
            f"Iteration {iteration + 1}: "
            f"Current = {current}, f(x) = {current_value}, "
            f"Best Neighbor = {best_neighbor}, f(x) = {best_neighbor_value}"
        )

        if best_neighbor_value <= current_value:
            break

        current = best_neighbor
        current_value = best_neighbor_value

    return current, current_value
```

### Step 4: Run Hill Climbing

Now we run the algorithm from an initial starting point.

Since the function is:

$$
f(x) = -x^2 + 6x
$$

its maximum is at:

$$
x = 3
$$

So the algorithm should move toward `x = 3`.


```python
best_x, best_value = hill_climbing(start=0, step_size=1)

print("-" * 30)
print(f"Best Solution Found: x = {best_x}")
print(f"Best Objective Value: {best_value}")
```

```python
best_x, best_value = hill_climbing(start=6, step_size=1)

print("-" * 30)
print(f"Best Solution Found: x = {best_x}")
print(f"Best Objective Value: {best_value}")
```

### Step 4: A More General Version

Below is a reusable version of Hill Climbing where we pass the objective function
as an argument. This is useful for solving other optimization problems.


```python
def hill_climbing_general(objective_fn, start, step_size=1, max_iterations=100):
    current = start
    current_value = objective_fn(current)

    for _ in range(max_iterations):
        neighbors = [current - step_size, current + step_size]
        best_neighbor = max(neighbors, key=objective_fn)
        best_neighbor_value = objective_fn(best_neighbor)

        if best_neighbor_value <= current_value:
            return current, current_value

        current = best_neighbor
        current_value = best_neighbor_value

    return current, current_value
```

```python
x, value = hill_climbing_general(objective_function, start=-2, step_size=1)

print(f"Best x: {x}")
print(f"Best value: {value}")
```
