```python
import heapq
```

### Uniform Cost Search (UCS)

Uniform Cost Search (UCS) is a search algorithm that expands the node with the **lowest cumulative path cost**.  
It is equivalent to Dijkstra’s algorithm for goal-based search and always finds the **optimal path** for any positive-cost graph.

#### Key Characteristics
- Expands the lowest-cost node first (priority queue)
- Guarantees optimality for positive edge costs
- Equivalent to Dijkstra’s shortest path algorithm
- Often more efficient than BFS on weighted graphs

#### Complexity
- Time: O(E log V)
- Space: O(V)

```python
def ucs_search(graph, start, goal):
    """
    Uniform Cost Search (Graph Search)
    graph: dict { node: [(neighbor, cost), ...] }
    """
    frontier = [(0, start, [start])]   # (cost, node, path)
    visited = {}

    while frontier:
        cost, node, path = heapq.heappop(frontier)

        # If this node was seen more cheaply before → skip
        if node in visited and visited[node] <= cost:
            continue

        visited[node] = cost

        if node == goal:
            return path, cost

        for (neighbor, edge_cost) in graph.get(node, []):
            new_cost = cost + edge_cost
            heapq.heappush(frontier, (new_cost, neighbor, path + [neighbor]))

    return None, float("inf")
```

```python
graph = {
    "A": [("B", 2), ("C", 8)],
    "B": [("D", 2), ("E", 5)],
    "C": [("F", 2)],
    "D": [("G", 5)],
    "E": [("G", 1)],
    "F": [("G", 1)],
    "G": []
}

path, cost = ucs_search(graph, "A", "G")
print("Best path:", path)
print("Total cost:", cost)
```
