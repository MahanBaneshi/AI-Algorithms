### Depth-Limited Search (DLS)

Depth-Limited Search (DLS) is a variant of DFS where the depth of exploration is restricted by a fixed limit.  
It prevents infinite descent in cyclic or deep graphs but may fail to find shallow solutions if the limit is too small.

#### Key Points
- DFS with a maximum exploration depth  
- Avoids infinite loops in deep graphs  
- May miss solutions if depth limit is insufficient  
- Two variants:
  - **Tree Search**: no visited set → may revisit states
  - **Graph Search**: uses visited set → avoids cycles

#### Complexity
- Time: O(b^ℓ)  
- Space: O(b * ℓ)  
Where **ℓ** is the depth limit.

```python
def dls_search(graph, start, goal, limit):
    """
    Depth-Limited Search (Graph Search version)
    Prevents cycles and revisits.
    """
    def recursive_dls(node, depth, path, visited):
        if node == goal:
            return path
        
        if depth == limit:
            return None
        
        visited.add(node)
        
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                result = recursive_dls(neighbor, depth + 1, path + [neighbor], visited)
                if result is not None:
                    return result
        
        visited.remove(node)
        return None
    
    return recursive_dls(start, 0, [start], set())
```

```python
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": ["G"],          
    "E": ["H", "I"],
    "F": [],             
    "G": [],             
    "H": ["J"],
    "I": [],             
    "J": ["K"],
    "K": ["Goal"],
    "Goal": ["E"]      
}
```

```python
print("limit=2:", dls_search(graph, "A", "Goal", 2))
print("limit=3:", dls_search(graph, "A", "Goal", 3))
print("limit=4:", dls_search(graph, "A", "Goal", 4))
print("limit=5:", dls_search(graph, "A", "Goal", 5))
print("limit=6:", dls_search(graph, "A", "Goal", 6))
print("limit=7:", dls_search(graph, "A", "Goal", 7))
```
