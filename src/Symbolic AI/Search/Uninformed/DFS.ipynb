### Depth-First Search (DFS)

DFS explores **deep paths first** using a **stack** (or recursion).  
It is memory-efficient but does **not** guarantee shortest paths.

**Key points:**
- Uses a **stack** (LIFO)
- Deep-first exploration
- Very low memory usage
- Tree Search can fall into cycles
- Graph Search avoids repeated states with a visited set

**Complexity:**
- Time: O(b^m)
- Space: O(b * m)

```python
def dfs_search(graph, start, goal):
    stack = [[start]]
    visited = set()

    while stack:
        path = stack.pop()
        node = path[-1]

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)

            for neighbor in reversed(graph.get(node, [])):
                new_path = list(path)
                new_path.append(neighbor)
                stack.append(new_path)

    return None
```

```python
graph = {
    'A': ['B', 'C', 'D'],
    'B': ['E', 'F', 'G'],
    'C': ['H'],
    'D': ['I'],
    'E': [],
    'F': ['J'],
    'G': ['K'],
    'H': ['K', 'L'],
    'I': ['M', 'N'],
    'J': [],
    'K': ['O', 'P'],
    'L': [],
    'M': [],
    'N': ['Q', 'R', 'S'],
    'O': [],
    'P': ['T', 'U'],
    'Q': [],
    'R': [],
    'S': [],
    'T': [],
    'U': []
}
print(dfs_search(graph, 'A', 'U'))
print(dfs_search(graph, 'A', 'R'))

```

#### BFS vs DFS (Quick Comparison)

- **BFS**: shortest path, high memory, level-order expansion  
- **DFS**: memory-efficient, deep exploration, not optimal  
- Both have Tree Search & Graph Search variants  
- Graph Search is necessary for cyclic graphs
