```python
from collections import deque
```

### Breadth-First Search (BFS)

BFS explores the search space **level by level** using a **queue**.  
It always finds the **shortest path** (in number of edges).

**Key points:**
- Uses a **queue** (FIFO)
- Guarantees shortest path
- Suitable for unweighted graphs
- Tree Search may revisit states
- Graph Search uses a visited set to avoid cycles

**Complexity:**
- Time: O(b^d)
- Space: O(b^d)

```python
def bfs_search(graph, start, goal):
    frontier = deque([[start]])
    explored = set()

    while frontier:
        path = frontier.popleft()
        node = path[-1]

        if node == goal:
            return path

        if node not in explored:
            explored.add(node)

            for neighbor in graph.get(node, []):
                new_path = list(path)
                new_path.append(neighbor)
                frontier.append(new_path)

    return None
```

```python
# test

def test_bfs():
    graph = {
        'A': ['B', 'C'],
        'B': ['A', 'D', 'E'],
        'C': ['A', 'F'],
        'D': ['B'],
        'E': ['B', 'F'],
        'F': ['C', 'E']
    }

    start_node = 'A'
    goal_node = 'F'
    
    result = bfs_search(graph, start_node, goal_node)
    
    if result:
        print(f"Path found from {start_node} to {goal_node}: {' -> '.join(result)}")
    else:
        print(f"No path found from {start_node} to {goal_node}.")

test_bfs()
```
