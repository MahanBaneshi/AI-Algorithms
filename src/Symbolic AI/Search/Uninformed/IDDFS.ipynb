### Iterative Deepening Depth-First Search (IDDFS)

IDDFS combines the advantages of **BFS** and **DFS**:

- Memory usage similar to DFS (very small)
- Completeness similar to BFS (guaranteed to find the shallowest goal)
- Performs DFS repeatedly with increasing depth limits

#### How it Works
1. Run DLS with limit = 0  
2. Run DLS with limit = 1  
3. Run DLS with limit = 2  
4. … until max depth is reached or goal is found

#### Complexity
- Time: O(b^d)  
- Space: O(b * d)  
Same optimality guarantees as BFS, but with DFS-level space complexity.

```python
def dls_search(graph, start, goal, limit):
    """Depth-Limited Search used internally by IDDFS."""
    def recursive_dls(node, depth, path, visited):
        if node == goal:
            return path
        
        if depth == limit:
            return None
        
        visited.add(node)
        
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                result = recursive_dls(neighbor, depth + 1, path + [neighbor], visited)
                if result is not None:
                    return result
        
        visited.remove(node)
        return None
    
    return recursive_dls(start, 0, [start], set())
```

```python
def iddfs_search(graph, start, goal, max_depth):
    """
    Iterative Deepening Depth-First Search
    Runs DLS from depth = 0 up to max_depth
    """
    for limit in range(max_depth + 1):
        result = dls_search(graph, start, goal, limit)
        if result is not None:
            return result, limit
    return None, None
```

```python
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": [],
    "D": ["F"],
    "E": [],
    "F": ["Goal"],
    "Goal": []
}

path, depth = iddfs_search(graph, "A", "Goal", max_depth=6)

print("Found path:", path)
print("Depth found:", depth)
```
