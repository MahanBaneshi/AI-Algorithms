# Alpha-Beta Pruning

Alpha-Beta pruning is an optimization technique for the minimax algorithm. 
It reduces the number of nodes evaluated by the minimax algorithm in its search tree.

### How it works
It maintains two values, **Alpha** and **Beta**:
- **Alpha ($\alpha$):** The best value that the **Maximizer** currently can guarantee. (Initially $-\infty$)
- **Beta ($\beta$):** The best value that the **Minimizer** currently can guarantee. (Initially $+\infty$)

If at any point $\beta \leq \alpha$, it means the current branch cannot possibly influence the final decision, and we can "prune" (skip) the rest of its children.


### Step 1: Node Representation
We use the same `GameNode` class as before to represent our game tree.


```python
class GameNode:
    def __init__(self, name, children=None, value=None):
        self.name = name
        self.children = children if children is not None else []
        self.value = value

    def is_terminal(self):
        return self.value is not None
```

### Step 2: Alpha-Beta Pruning Implementation
This function implements the core logic. Notice how we pass `alpha` and `beta` down the tree 
and break the loop if a pruning condition is met.


```python
def alpha_beta(node, alpha, beta, maximizing_player):
    if node.is_terminal():
        return node.value

    if maximizing_player:
        value = float('-inf')
        for child in node.children:
            value = max(value, alpha_beta(child, alpha, beta, False))
            alpha = max(alpha, value)
            if beta <= alpha:
                print(f"Pruning at node: {node.name} (Max Player)")
                break  # Beta cut-off
        return value
    else:
        value = float('inf')
        for child in node.children:
            value = min(value, alpha_beta(child, alpha, beta, True))
            beta = min(beta, value)
            if beta <= alpha:
                print(f"Pruning at node: {node.name} (Min Player)")
                break  # Alpha cut-off
        return value
```

### Step 3: Create a Deeper Tree to See Pruning in Action
To demonstrate pruning, we need a tree where certain branches are guaranteed 
not to be picked.


```python
# Level 3 (Leaves)
n1 = GameNode("n1", value=3)
n2 = GameNode("n2", value=5)
n3 = GameNode("n3", value=6)
n4 = GameNode("n4", value=9)
n5 = GameNode("n5", value=1)
n6 = GameNode("n6", value=2)
n7 = GameNode("n7", value=0)
n8 = GameNode("n8", value=-1)

# Level 2 (Min Nodes)
min_node_1 = GameNode("Min1", children=[n1, n2])
min_node_2 = GameNode("Min2", children=[n3, n4])
min_node_3 = GameNode("Min3", children=[n5, n6])
min_node_4 = GameNode("Min4", children=[n7, n8])

# Level 1 (Max Nodes)
max_node_1 = GameNode("Max1", children=[min_node_1, min_node_2])
max_node_2 = GameNode("Max2", children=[min_node_3, min_node_4])

# Root (Min Player)
root = GameNode("Root", children=[max_node_1, max_node_2])

result = alpha_beta(root, float('-inf'), float('inf'), False)

print("-" * 20)
print(f"Final Alpha-Beta Value: {result}")
```
