# MiniMax Algorithm

MiniMax is a decision-making algorithm used in **two-player adversarial games** such as
Tic-Tac-Toe, Chess, and Checkers.

It assumes:
- One player is the **maximizer** and tries to maximize the score.
- The other player is the **minimizer** and tries to minimize the score.

### Main Idea
MiniMax explores the game tree recursively:
- At **MAX** nodes, it chooses the move with the **highest** value.
- At **MIN** nodes, it chooses the move with the **lowest** value.

### Terminal States
The recursion stops when:
- the game is over, or
- a specified depth limit is reached.

At that point, the algorithm returns:
- the true game result, or
- a heuristic evaluation score.

MiniMax guarantees the optimal move if the full game tree is explored.

### Step 1: Game Tree Node Representation

We define a simple tree node structure for a game tree.
Each node stores:
- a name
- an optional score (for terminal nodes)
- a list of child nodes


```python
class GameNode:
    def __init__(self, name, children=None, value=None):
        self.name = name
        self.children = children if children is not None else []
        self.value = value

    def is_terminal(self):
        return self.value is not None
```

### Step 2: MiniMax Implementation

This function recursively evaluates the game tree:
- If the node is terminal, return its value.
- If it is a MAX turn, return the maximum value among children.
- If it is a MIN turn, return the minimum value among children.


```python
def minimax(node, maximizing_player):
    if node.is_terminal():
        return node.value

    if maximizing_player:
        best_value = float('-inf')
        for child in node.children:
            value = minimax(child, False)
            best_value = max(best_value, value)
        return best_value
    else:
        best_value = float('inf')
        for child in node.children:
            value = minimax(child, True)
            best_value = min(best_value, value)
    return best_value
```

### Step 3: Build an Example Game Tree

We define a small game tree manually.

Tree structure:

A(MAX)
/        \
B(MIN)      C(MIN)
/   |   \    /   |   \
3    5   2   9   12   5

MiniMax should evaluate:
- node B as min(3, 5, 2) = 2
- node C as min(9, 12, 5) = 5
- node A as max(2, 5) = 5


```python
leaf1 = GameNode("L1", value=3)
leaf2 = GameNode("L2", value=5)
leaf3 = GameNode("L3", value=2)
leaf4 = GameNode("L4", value=9)
leaf5 = GameNode("L5", value=12)
leaf6 = GameNode("L6", value=5)

node_b = GameNode("B", children=[leaf1, leaf2, leaf3])
node_c = GameNode("C", children=[leaf4, leaf5, leaf6])

root = GameNode("A", children=[node_b, node_c])
result = minimax(root, True)
print("MiniMax Value:", result)

```

### Step 4: Choose the Best Move from the Root

Sometimes we want not only the final MiniMax value, but also the best child node
to choose from the root.

```python
def best_move(node):
    best_child = None
    best_value = float('-inf')

    for child in node.children:
        value = minimax(child, False)
        print(f"Move {child.name} => value {value}")
        if value > best_value:
            best_value = value
            best_child = child

    return best_child, best_value

move, value = best_move(root)
print("Best Move:", move.name)
print("Best Value:", value)
```
