# Greedy Best-First Search (GBFS)

Greedy Best-First Search is a heuristic search algorithm that always expands the node
that appears to be closest to the goal according to a heuristic function `h(n)`.

### Key Idea
- Uses only the heuristic value
- Chooses the most promising node first
- Very fast in many cases
- Does not guarantee the shortest or cheapest path

### Main Components
- `graph`: adjacency list of the search space
- `heuristic`: estimated distance from each node to the goal
- priority queue: stores nodes ordered by heuristic value

### Step 1: Path Reconstruction
This helper function rebuilds the final path after the goal is found,
using the parent pointers recorded during the search.


```python
def reconstruct_path(parents, start, goal):
    path = [goal]
    current = goal

    while current != start:
        current = parents[current]
        path.append(current)

    path.reverse()
    return path
```

### Step 2: Greedy Best-First Search
This function uses a priority queue ordered only by heuristic values.
The node with the smallest `h(n)` is expanded first.


```python
import heapq
from itertools import count

def greedy_best_first_search(graph, heuristic, start, goal):
    if start == goal:
        return [start]

    frontier = []
    tie_breaker = count()

    # (heuristic value, tie-breaker, node)
    heapq.heappush(frontier, (heuristic[start], next(tie_breaker), start))

    visited = set()
    parents = {start: None}

    while frontier:
        h, _, node = heapq.heappop(frontier)

        if node in visited:
            continue

        visited.add(node)

        if node == goal:
            return reconstruct_path(parents, start, goal)

        for neighbor in graph.get(node, []):
            if neighbor not in visited and neighbor not in parents:
                parents[neighbor] = node
                heapq.heappush(frontier, (heuristic[neighbor], next(tie_breaker), neighbor))

    return None
```

### Step 3: Example Graph and Heuristic
The graph below is designed so that the heuristic guides the search toward the goal.
Notice that GBFS chooses the node that *looks* closest, not necessarily the cheapest path.


```python
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F", "G"],
    "D": ["H"],
    "E": ["I", "J"],
    "F": [],
    "G": ["K"],
    "H": [],
    "I": [],
    "J": [],
    "K": ["Goal"],
    "Goal": []
}

heuristic = {
    "A": 7,
    "B": 6,
    "C": 2,
    "D": 5,
    "E": 4,
    "F": 3,
    "G": 1,
    "H": 6,
    "I": 3,
    "J": 2,
    "K": 1,
    "Goal": 0
}
path = greedy_best_first_search(graph, heuristic, "A", "Goal")
print("Found path:", path)

```
