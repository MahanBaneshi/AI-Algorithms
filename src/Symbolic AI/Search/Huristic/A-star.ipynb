# A* Search Algorithm

A* Search is one of the most popular and effective search algorithms. It is **optimal** and **complete** (provided the heuristic is admissible and consistent).

### How it Works
A* uses a evaluation function $f(n) = g(n) + h(n)$:
- **$g(n)$**: The actual cost to reach node $n$ from the start.
- **$h(n)$**: The estimated cost to reach the goal from node $n$.
- **$f(n)$**: The total estimated cost of the path through node $n$.

### Features
- Guaranteed to find the shortest path (if $h(n)$ is admissible).
- More efficient than Uniform Cost Search because it is guided by the heuristic.


### Step 1: Path Reconstruction
A simple helper to backtrack from the goal to the start using the parent dictionary
recorded during the search.


```python
def reconstruct_path(parents, start, goal):
    path = [goal]
    current = goal
    while current != start:
        current = parents[current]
        path.append(current)
    path.reverse()
    return path

```

### Step 2: A* Implementation
This function uses a priority queue to always expand the node with the lowest $f(n)$.
It keeps track of the best $g(n)$ found so far for each node to avoid suboptimal paths.


```python
import heapq
from itertools import count

def a_star_search(graph, heuristic, start, goal):
    if start == goal:
        return [start], 0

    # Priority Queue: (f_score, tie_breaker, current_node)
    frontier = []
    tie_breaker = count()
    
    # f(start) = g(start) + h(start) = 0 + h(start)
    heapq.heappush(frontier, (heuristic[start], next(tie_breaker), start))
    
    parents = {start: None}
    g_score = {start: 0}
    
    while frontier:
        f, _, current = heapq.heappop(frontier)
        
        if current == goal:
            return reconstruct_path(parents, start, goal), g_score[goal]
        
        # graph format: {node: [(neighbor, cost), ...]}
        for neighbor, cost in graph.get(current, []):
            tentative_g_score = g_score[current] + cost
            
            # If this path to neighbor is better than any previous one
            if neighbor not in g_score or tentative_g_score < g_score[neighbor]:
                parents[neighbor] = current
                g_score[neighbor] = tentative_g_score
                f_score = tentative_g_score + heuristic.get(neighbor, float('inf'))
                heapq.heappush(frontier, (f_score, next(tie_breaker), neighbor))
                
    return None, float('inf')
```

### Step 3: Example Weighted Graph and Heuristic
For A*, the graph must include edge weights (costs). 
We also provide a heuristic dictionary representing the estimated distance to the goal.


```python
graph = {
'Start': [('A', 5), ('B', 8), ('C', 12)],
'A': [('D', 6), ('E', 15)],
'B': [('E', 4), ('F', 10)],
'C': [('G', 3)],
'D': [('H', 7)],
'E': [('H', 2), ('I', 5)],
'F': [('I', 8), ('J', 12)],
'G': [('J', 6)],
'H': [('K', 5), ('Goal', 15)],
'I': [('K', 3), ('Goal', 7)],
'J': [('Goal', 4)],
'K': [('Goal', 2)],
}

heuristic = {
'Start': 20,
'A': 16,
'B': 14,
'C': 18,
'D': 12,
'E': 8,
'F': 10,
'G': 10,
'H': 6,
'I': 4,
'J': 3,
'K': 1,
'Goal': 0
}

path, cost = a_star_search(graph, heuristic, 'Start', 'Goal')
print("Optimal Path:", path)
print("Total Path Cost:", cost)
```
